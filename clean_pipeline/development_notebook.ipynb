```python
from color_assigner import ColorAssigner
from custom_types import FrameDetections
import json

color_assigner = ColorAssigner()

with open(f"yolo_player_tracker_results/{input_video.split('.mp4')[0]}/player_detections.json", "r") as f:
    raw_detections = json.load(f)
detections: list[FrameDetections] = [FrameDetections.model_validate(frame_detections) for frame_detections in raw_detections]


detections: list[FrameDetections] = color_assigner.process_video(input_path=f"{input_folder}/{input_video}", detections=detections)
```

```python
from coordinate_transformer import CoordinateTransformer
from custom_types import FrameDetections
import json

coordinate_transformer = CoordinateTransformer()

with open(f"color_assigner_results/{input_video.split('.mp4')[0]}/player_colors.json", "r") as f:
    raw_detections = json.load(f)
detections: list[FrameDetections] = [FrameDetections.model_validate(frame_detections) for frame_detections in raw_detections]

detections = coordinate_transformer.transform(input_path=f"{input_folder}/{input_video}", detections=detections)
```

```python
from role_assigner2 import RoleAssigner
from custom_types import FrameDetections
import json

input_folder = "input"
input_video = "SNGS-007_video_cut.mp4"

role_assigner = RoleAssigner()

with open(f"coordinate_transformer_results/{input_video.split('.mp4')[0]}/detections_with_minimap.json", "r") as f:
    raw_detections = json.load(f)
detections: list[FrameDetections] = [FrameDetections.model_validate(frame_detections) for frame_detections in raw_detections]

detections = role_assigner.assign_roles(input_path=f"{input_folder}/{input_video}", detections=detections)
```
